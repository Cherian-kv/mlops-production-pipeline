```python
import torch
import mlflow
```

```python
# 1. Test the Conda Foundation (CUDA)
print(f"PyTorch Version: {torch.__version__}")
cuda_available = torch.cuda.is_available()
print(f"CUDA Available:  {cuda_available}")

if cuda_available:
    device = torch.device("cuda")
    # Force the GPU to do some actual math
    tensor = torch.rand(5, 5).to(device)
    result = tensor @ tensor
    print(f"GPU Math Test:   SUCCESS on {result.device}")
else:
    print("GPU Math Test:   FAILED. CUDA is unreachable.")
    exit()
```

```python
print("\n--- MLOPS CHECK ---")

# 2. Test the Pip Addition (MLflow)
print(f"MLflow Version:  {mlflow.__version__}")

# Tell MLflow to create a local tracking database
mlflow.set_experiment("system_survival_test")

with mlflow.start_run():
    # Log some dummy data just to prove the database writes correctly
    mlflow.log_param("hardware", "NVIDIA_GPU")
    mlflow.log_metric("matrix_sum", float(result.sum()))
    print("MLflow Tracking: SUCCESS. Data written to local tracking server.")
```
